# encoding3d scenes

[encoding3d](https://encoding3d.readthedocs.io) is an exchange format for whole functional-structural plant model
(FSPM) scenes, based on glTF: multiscale plant graphs, their variables, their geometry and the environment's grids,
one file per time step. `openalea.widgets.encoding3d` shows them in the notebook with k3d, colored by a variable.

The scenes of this notebook are small reference scenes of the encoding3d repository (its conformance files).
It needs `openalea.encoding3d`:

```bash
mamba install -c openalea3 -c conda-forge openalea.widgets openalea.encoding3d
```

````python
from openalea import encoding3d
from openalea.widgets.encoding3d import Encoding3D, variables
````

## A root system, colored by a variable

`Encoding3D` takes a frame, or the path of its `.glb` file, and the variable coloring the meshes. The vertices of a
mesh refer to their entity at each scale (here the root segments, the axes and the plant): the variable is taken
from the first one that has it, or from `entity_set`.

````python
Encoding3D("encoding3d/root_system.glb", variable="radius")
````

The variables that can color the meshes, by entity set:

````python
variables(encoding3d.read("encoding3d/root_system.glb"))
````

````python
Encoding3D("encoding3d/root_system.glb", variable="branching_order", entity_set="Axis")
````

## A wheat shoot

The leaf parts colored by the PAR they absorb.

````python
Encoding3D("encoding3d/wheat_shoot.glb", variable="PARa")
````

## Roots in the soil

A grid of the environment (here the soil) is shown as translucent cells, colored by `grid_variable`. The legend is
the one of the plants' variable.

````python
Encoding3D("encoding3d/soil_and_roots.glb", variable="radius", grid_variable="soil_water_potential")
````

## Anatomy: the cells of a root section

````python
Encoding3D("encoding3d/section_cells.glb", variable="water_potential")
````

## A growing root system: a sequence of frames

A sequence (its index `scene.fspm.json`, or its folder) gives a plot with one time step per frame: play it with
the time slider of the k3d panel (top right of the view). The colors share one range over the frames.

````python
Encoding3D("encoding3d/growing_sequence", variable="length")
````
